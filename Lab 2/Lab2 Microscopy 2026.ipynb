# Introduction

<font color='black'> As we discussed in the class, the propagation of waves can be conceptualized by Fresnel-Huygens’ principle that the integral of spherical wavelets from primary wavefronts will describe how waves propagate. We learned about Fresnel and Fraunhofer diffraction by different mathematical approximations.
    <br></br>
When the wave interacts with a positive lens, the lens provides a phase transformation as a function of (x, y) in the transverse plane to effectively cancel the phase delay caused by the wave propagation (i.e. a paraboloidal wave in Fresnel diffraction by a paraxial approximation). When the image target, O, is placed one focal length away from a lens, the reciprocal plane, P, on the other side of the lens at exactly one focal length away, will be a Fourier transform of the object. If we place another lens one focal length away from P, the second lens performs another Fourier transform and creates a real image, I, of the target O. This is a classic 4f system (literally 4 focal lengths). Note that the image orientation will be upside down, left side right to the object. This makes sense mathematically if you think about the result of two identical Fourier transform in a row.
        <br></br> 
This lab focuses on observing and characterizing wave propagation and diffraction, and Fourier transform properties of a lens. Additionally, we will need to setup the system initially, during which some of the geometric optics will be practiced.

**Answer:**

1. Estimate the Fresnel Diffraction region and Fraunhofer region

As we can see from the images from the lab session, as the distance from the target increases, the diffraction pattern transitions from the Fresnel (near-field) region to the Fraunhofer (far-field) region. In the Fresnel region, the grid structure remains recognizable and resembles a projection of the original target (50µm grid pattern from the R1L1S1N resolution target). In contrast, the Fraunhofer pattern represents the Fourier transform of the grid, showing regularly spaced bright diffraction orders in both the horizontal (x) and vertical (y) directions. These diffraction orders form a cross-shaped pattern as shown in the figures above.

For our experiment, a 532 nm laser was directed through a grid pattern on the Thorlabs R1L1S1N resolution target, where the diffraction patterns were observed at different distances from the target. We can use the Fresnel number to estimate the near-field and far-field diffraction regions:

$N_F = a^2 / (\lambda z)$

where a is the half-width of the illuminated aperture, $\lambda$ is the laser wavelength, and z is the distance from the target to the observation plane.

For a 50 µm-wide opening, a = 25 µm. Using the 532 nm laser wavelength, the Fresnel number becomes:

$N_F = \frac{1.175}{z}$

When N_F is much greater than 1, the observation plane is in the Fresnel (near-field) region, and when N_F is much smaller than 1, it approaches the Fraunhofer (far-field) region. The transition occurs approximately when $N_F = 1$, corresponding to a propagation distance of 1.175 mm for this aperture.

2. Estimate the period of the grid pattern and the width of each slit.

For our experiment, $\lambda = 532$nm, $\Delta x = 4$ mm, and $z = 37.6$ cm. Using these numbers, we can estimate the grid period using the following equation: 

$$
\begin{align}

p &= \frac{\lambda z}{\Delta x'} \\
&= \frac{532 * 10^{-9} * 0.376}{0.004} \\
&= 50µm

\end{align}
$$

Thus, the estimated grid period is approximately 50µm. 

Next, using the equation below at the first minimum distance, we can estimate the width of each slit to be:

$$
\begin{align}
a &= \frac{\lambda z }{x_{min}'}
&= \frac{(532 * 10^{-9} * 0.376)} {0.00444}
&= 45 µm
\end{align}
$$

Thus, the width of the transparent slit is approximately 45 µm, assuming each period contains a 5 µm opaque line.

## Line Spread Function (LSF)
The Line Spread Function (LSF) describes how an optical system responds to a line source of light. Essentially, it represents the redistribution of photons from a source that is a delta function, both spectrally and spatially, integrated in either the dispersion or cross-dispersion direction.

In simpler terms, the LSF is derived by integrating the point spread function (PSF) along sections parallel to the direction of the line. This is because a line image is the summation of an infinite number of image points along its length
