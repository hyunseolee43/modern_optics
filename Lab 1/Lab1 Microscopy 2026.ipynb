<b> Q: What changes do you observe in the position of the light on the target? Explain using your knowledge of geometric optics </b>

A: The position of the light on the target moves in the same direction as the lens when it is moved laterally. When the beam passes through the center of the lens, it is not deflected and the spot returns to its original, pre-lens position. We used this knowledge to center the system laterally.

We used **USAF 1951 Group 2 Element 2 (G2E2)** to measure the magnification. G2E2 has a linewidth of approximately **111 µm**. The three bars and two spaces of G2E2 span 5 line widths:

$$
5(111\,\mu\text{m})=555\,\mu\text{m}=0.555\text{ mm}.
$$

In the image, these 5 lines measured approximately **4 mm**, giving a measured magnification of

$$
M=\frac{4}{0.555}\approx \mathbf{7.2\times}.
$$

Experimentally, the object distance for lens 1 was **50 mm**, the lens separation was **250 mm**, and the final image distance $d_{i2}$ was approximately **855 mm**.

For lens 1,

$$
\frac{1}{30}=\frac{1}{50}+\frac{1}{d_{i1}},
$$

giving

$$
d_{i1}=75\text{ mm}.
$$

The magnification of lens 1 is

$$
m_1=-\frac{75}{50}=-1.5.
$$

Since the lens separation is 250 mm, the object distance for lens 2 is

$$
d_{o2}=250-75=175\text{ mm}.
$$

For lens 2,

$$
\frac{1}{150}=\frac{1}{175}+\frac{1}{d_{i2}},
$$

giving

$$
d_{i2}=1050\text{ mm}.
$$

The magnification of lens 2 is

$$
m_2=-\frac{1050}{175}=-6.
$$

Therefore, the theoretical total magnification is

$$
M=m_1m_2=(-1.5)(-6)=\mathbf{9\times}.
$$

| Quantity | Measured | Calculated | Error |
|---|---:|---:|---:|
| Magnification | 7.2× | 9.0× | 20.0% |
| Final image distance | 855 mm | 1050 mm | 18.6% |

Thus, based on the measured **50 mm object distance** and **250 mm lens separation**, the theoretical magnification is **9×** and the theoretical final image distance is **1050 mm**.

The ~20% differences are larger than in the 4f measurement, and this is expected for this geometry. Lens 2's object distance (175 mm) is close to its focal length (150 mm), so the image distance is very sensitive to small errors: the axial magnification of lens 2 is (d_i2/d_o2)² = 36, so a 1 mm error in locating the intermediate image shifts the final image by ~36 mm. For example, if the object were 53 to 54 mm from lens 1 instead of 50 mm, the predicted final image distance would drop to roughly the measured 855 mm. Other sources of error include measuring distances to lens mounts rather than the lens principal planes, and reading the 4 mm image width by eye.

This highlights the difference between finite and infinite conjugate imaging. In the 4f system, the object sat at f1, the light between the lenses was collimated, and the magnification was fixed at -f2/f1 regardless of the lens spacing. Here (finite conjugate), the light between the lenses converges to a real intermediate image, so the final image distance and magnification depend on both the object distance and the lens separation, and small changes in either produce large changes in the result.
